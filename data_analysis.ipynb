# 📊 Data Analysis with NumPy, Pandas, and Matplotlib

This notebook demonstrates the basics of data manipulation, cleaning, exploration, and visualization using Python's standard data science library stack: **NumPy**, **Pandas**, and **Matplotlib**.

---

## 🛠️ Step 1: Import Libraries
First, we import the required libraries:
- **NumPy**: For numerical computing and working with arrays.
- **Pandas**: For data structures and data analysis (DataFrames).
- **Matplotlib**: For creating static, interactive, and animated visualizations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Print versions to verify
print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {pd.__version__}")
```

## 🔢 Step 2: Working with NumPy
Let's practice some core NumPy operations, including array creation, statistical calculations, and indexing.

```python
# Create a 1D NumPy array representing daily sales values
sales = np.array([120, 150, 85, 210, 130, 175, 250])
print("Daily Sales Array:", sales)

# Basic statistics using NumPy
print("Average Daily Sales:", np.mean(sales))
print("Median Sales:", np.median(sales))
print("Standard Deviation:", np.std(sales))
print("Highest Sales Day:", np.max(sales))

# Filtering arrays: Sales above 150
high_sales = sales[sales > 150]
print("Days with sales > 150:", high_sales)
```

## 🐼 Step 3: Loading and Exploring a Dataset with Pandas
We will load the **Tips Dataset** directly from a public URL. This dataset contains information about tips received by a food server over a period of time.

```python
# Load dataset from Seaborn's public data source
dataset_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(dataset_url)

# Display the first 5 rows of the dataset
df.head()
```

### Exploring the Dataset Structure
Let's check the shape, data types, and general information about our dataset.

```python
print("--- Dataset Shape ---")
print(df.shape)  # Rows, Columns

print("\n--- Dataset Info & Data Types ---")
df.info()

print("\n--- Descriptive Statistics ---")
df.describe()
```

## 🧹 Step 4: Data Cleaning and Analysis
Let's check if there are any missing (null) values in our dataset and perform simple data aggregation.

```python
# Check for null values
print("Missing values in each column:")
print(df.isnull().sum())

# Let's perform a simple grouping operation:
# Group by day and calculate the average total bill and tip size
grouped_by_day = df.groupby('day')[['total_bill', 'tip']].mean()
print("\nAverage Bill and Tip by Day:")
print(grouped_by_day)
```

## 📈 Step 5: Data Visualization with Matplotlib
We will now create three types of charts:
1. **Bar Chart**: Showing average total bill by day.
2. **Line Chart**: Showing the tip amount vs. total bill to explore patterns.
3. **Pie Chart**: Showing the proportion of total customers by meal time (Lunch vs. Dinner).

```python
# 1. Bar Chart: Average Total Bill by Day
plt.figure(figsize=(8, 5))
days = grouped_by_day.index
avg_bills = grouped_by_day['total_bill']

# Use a premium color palette (e.g. shades of teal)
plt.bar(days, avg_bills, color='#2c7a7b', edgecolor='black', width=0.6)

# Formatting
plt.title('Average Total Bill by Day', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Bill ($)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the chart
plt.show()
```

```python
# 2. Line Chart: Tip Amount vs. Total Bill (Sorted by Bill)
plt.figure(figsize=(10, 5))

# Sort by total bill for a cleaner line representation
df_sorted = df.sort_values(by='total_bill')

plt.plot(df_sorted['total_bill'], df_sorted['tip'], color='#e53e3e', marker='o', markersize=3, linestyle='-', linewidth=1, alpha=0.7)

# Formatting
plt.title('Trend of Tips by Total Bill Amount', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Bill ($)', fontsize=12)
plt.ylabel('Tip Amount ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
# 3. Pie Chart: Proportion of Customers by Time of Day (Lunch vs Dinner)
time_counts = df['time'].value_counts()

plt.figure(figsize=(6, 6))
colors = ['#3182ce', '#dd6b20'] # premium blue & orange

plt.pie(time_counts, labels=time_counts.index, autopct='%1.1f%%', startangle=140, colors=colors, 
        wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'antialiased': True}, 
        textprops={'fontsize': 12, 'fontweight': 'bold'})

plt.title('Customer Distribution by Meal Time', fontsize=14, fontweight='bold', pad=15)
plt.show()
```
