# 🤖 Machine Learning Fundamentals: Iris Flower Classification

This notebook walks through the basic **Machine Learning Workflow** using **Scikit-learn**, a popular machine learning library in Python.

## 📋 The Machine Learning Workflow:
1. **Define the Problem & Load the Dataset**: Understand the variables we are working with.
2. **Exploratory Data Analysis (EDA)**: Understand relationships in the data.
3. **Data Splitting**: Split the data into Training and Testing sets.
4. **Train a Model**: Fit a Machine Learning classifier to our training data.
5. **Evaluate the Model**: Test the model's performance on unseen data.
6. **Make Predictions**: Test the model with custom inputs.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning modules from Scikit-Learn
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries imported successfully!")
```

## 🗂️ Step 1: Load and Explore the Dataset
We'll use the classic **Iris Flower Dataset**. It contains 150 samples of three species of Iris flowers:
- **Setosa**
- **Versicolor**
- **Virginica**

Each flower is described by 4 features (measurements in centimeters):
1. Sepal Length
2. Sepal Width
3. Petal Length
4. Petal Width

```python
# Load dataset
iris = load_iris()

# Create a Pandas DataFrame for easier analysis
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Map species integers to actual names
df['species_name'] = df['species'].map({i: name for i, name in enumerate(iris.target_names)})

# Preview the data
df.head()
```

## 📊 Step 2: Exploratory Data Analysis (EDA)
Let's visualize the features to see if there are clear distinctions between the three species of flowers.

```python
# Pairplot to see relationships between features
plt.figure(figsize=(10, 8))
sns.pairplot(df.drop(columns=['species']), hue='species_name', palette='Set1')
plt.show()
```

## ✂️ Step 3: Splitting Data into Train & Test Sets
To make sure our model generalizes well to new data, we split our features ($X$) and targets ($y$) into:
- **Training Set (80%)**: Used to teach the model.
- **Testing Set (20%)**: Used to evaluate the model.

```python
# Define features and labels
X = iris.data
y = iris.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print(f"Training features shape: {X_train.shape}")
print(f"Testing features shape: {X_test.shape}")
```

## 🧠 Step 4: Training our Machine Learning Model
We will use a **Random Forest Classifier**, which is a powerful ensemble method that builds multiple decision trees and merges them to get a more accurate and stable prediction.

```python
# Initialize the model
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train (fit) the model
model.fit(X_train, y_train)

print("Model training complete!")
```

## 📏 Step 5: Evaluate the Model
We evaluate our model by comparing the predictions on the testing dataset (`X_test`) against the actual ground-truth labels (`y_test`).

```python
# Make predictions on test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"✅ Model Accuracy: {accuracy * 100:.2f}%\n")

# Detailed classification report
print("--- Classification Report ---")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, cmap='Blues', fmt='g', 
            xticklabels=iris.target_names, 
            yticklabels=iris.target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

## 🔮 Step 6: Make Predictions on Custom Input
Let's see if we can use our model to predict the species of a new, unseen flower sample.

```python
# Define measurements of a hypothetical flower
# [sepal length, sepal width, petal length, petal width]
new_flower = np.array([[5.1, 3.5, 1.4, 0.2]]) 

# Predict the species
prediction = model.predict(new_flower)
predicted_name = iris.target_names[prediction[0]]

print(f"Flower measurements: {new_flower[0]}")
print(f"🔮 Predicted Species: {predicted_name.upper()}")
```
