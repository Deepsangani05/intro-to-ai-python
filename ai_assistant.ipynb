# 🚀 AI-Powered Productivity Assistant

This notebook demonstrates how to build a simple **AI-powered assistant** in Python that integrates with the **Google Gemini API** (using the `google-generativeai` package) and includes a fallback **AI Simulator** if you don't have an API key handy.

## 🛠️ Step 1: Setting up the environment
We will install and import the required libraries. To interact with the real Gemini API, you need the `google-generativeai` library.

```python
# Install the official Google Generative AI package (if not already installed)
!pip install -q google-generativeai

import google.generativeai as genai
import os
print("Setup complete!")
```

## 🤖 Step 2: Defining the AI Simulator
If you run this notebook without a Gemini API Key, the **AI Simulator** will take over and mock responses for general chat, text summarization, and sentiment analysis.

```python
def simulate_ai_response(prompt, mode):
    """Simulates basic AI capabilities without an API key."""
    prompt_lower = prompt.lower()
    
    if mode == "chat":
        if "hello" in prompt_lower or "hi" in prompt_lower:
            return "AI (Simulator): Hello there! How can I help you learn AI today?"
        elif "python" in prompt_lower:
            return "AI (Simulator): Python is the standard language for AI and ML due to libraries like NumPy, Pandas, and Scikit-Learn!"
        else:
            return f"AI (Simulator): Interesting query! ('{prompt}'). Add your Gemini API key to get a real response from the AI model."
            
    elif mode == "summarize":
        words = prompt.split()
        summary = " ".join(words[:min(8, len(words))]) + "..."
        return f"AI (Simulator) [Summary]: {summary} (Reduced from {len(words)} words)"
        
    elif mode == "sentiment":
        positives = ["good", "great", "happy", "love", "excellent", "awesome"]
        negatives = ["bad", "sad", "hate", "terrible", "worst", "poor"]
        
        pos_score = sum(1 for w in positives if w in prompt_lower)
        neg_score = sum(1 for w in negatives if w in prompt_lower)
        
        sentiment = "POSITIVE 😊" if pos_score > neg_score else ("NEGATIVE 😞" if neg_score > pos_score else "NEUTRAL 😐")
        return f"AI (Simulator) [Sentiment]: {sentiment} (Keyword Matches: Pos={pos_score}, Neg={neg_score})"
        
    return "Unknown Mode"
```

## 🔌 Step 3: Run the Interactive AI Assistant
Run the cell below. It will ask for your Gemini API key. If you don't have one, just leave it blank and press Enter to run the application in **Simulator Mode**.

```python
# 1. Prompt for API Key
api_key = input("Enter your Google Gemini API Key (press Enter to skip & use Simulator): ").strip()

use_gemini = False
model = None

if api_key:
    try:
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel('gemini-pro')
        use_gemini = True
        print("✅ Real Google Gemini API initialized successfully!")
    except Exception as e:
        print(f"❌ Failed to initialize Gemini: {e}. Falling back to Simulator.")
else:
    print("ℹ️ Running in Simulator Mode (offline).")

# 2. App Menu Loop
print("\n" + "=" * 50)
print("   Welcome to your AI Productivity Assistant!   ")
print("=" * 50)

while True:
    print("\nSelect a task:")
    print("1. 💬 Chat with AI")
    print("2. 📝 Summarize Text")
    print("3. 📊 Sentiment Analysis")
    print("4. ❌ Exit")
    
    choice = input("Choice (1-4): ").strip()
    
    if choice == '4':
        print("Goodbye! 👋")
        break
        
    if choice not in ['1', '2', '3']:
        print("Invalid choice, please select again.")
        continue
        
    text_input = input("\nEnter your input text/prompt: ").strip()
    if not text_input:
        print("Input cannot be empty!")
        continue
        
    # Process based on mode and availability of Gemini API
    if choice == '1':
        if use_gemini:
            print("AI is thinking...")
            response = model.generate_content(text_input)
            print(f"\n🤖 Gemini: {response.text}")
        else:
            print(f"\n{simulate_ai_response(text_input, 'chat')}")
            
    elif choice == '2':
        if use_gemini:
            print("AI is summarizing...")
            response = model.generate_content(f"Summarize this text briefly: {text_input}")
            print(f"\n📝 Summary:\n{response.text}")
        else:
            print(f"\n{simulate_ai_response(text_input, 'summarize')}")
            
    elif choice == '3':
        if use_gemini:
            print("AI is analyzing sentiment...")
            response = model.generate_content(f"Analyze the sentiment of this text (Positive, Negative, or Neutral) and explain in 1 sentence: {text_input}")
            print(f"\n📊 Sentiment Analysis:\n{response.text}")
        else:
            print(f"\n{simulate_ai_response(text_input, 'sentiment')}")
```
