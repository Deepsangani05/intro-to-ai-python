# 🧮 Python Mini Project: Interactive Calculator

This is a simple, interactive Python calculator designed for the **AI & Python Internship**. 

### Features:
- **Basic Operations**: Addition, Subtraction, Multiplication, Division
- **Advanced Operations**: Exponentiation ($x^y$), Square Root ($\sqrt{x}$), Modulo ($x \% y$)
- **History**: Keeps track of all calculations performed in the current session
- **Error Handling**: Gracefully handles division by zero and non-numeric inputs

```python
import math

def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        raise ZeroDivisionError("Error: Division by zero is not allowed.")
    return x / y

def exponent(x, y):
    return x ** y

def square_root(x):
    if x < 0:
        raise ValueError("Error: Cannot calculate square root of a negative number.")
    return math.sqrt(x)

def modulo(x, y):
    if y == 0:
        raise ZeroDivisionError("Error: Modulo by zero is not allowed.")
    return x % y
```

```python
# Interactive Loop
history = []

print("=" * 40)
print("       ✨ INTERACTIVE CALCULATOR ✨       ")
print("=" * 40)
print("Operations: +, -, *, /, ^, sqrt, % | History: 'h' | Exit: 'q'")
print("=" * 40)

while True:
    choice = input("Enter operation (+, -, *, /, ^, sqrt, %, h, q): ").strip().lower()
    
    if choice == 'q':
        print("\nCalculator closed. Goodbye! 👋")
        break
        
    if choice == 'h':
        print("\n--- 📜 Calculation History ---")
        if not history:
            print("No calculations performed yet.")
        else:
            for idx, record in enumerate(history, 1):
                print(f"{idx}. {record}")
        print("=" * 40)
        continue
        
    if choice not in ['+', '-', '*', '/', '^', 'sqrt', '%']:
        print("❌ Invalid operation! Please try again.")
        continue
        
    try:
        if choice == 'sqrt':
            num = float(input("Enter the number: "))
            res = square_root(num)
            operation_str = f"√({num}) = {res}"
            print(f"Result: {operation_str}")
            history.append(operation_str)
        else:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            
            if choice == '+':
                res = add(num1, num2)
            elif choice == '-':
                res = subtract(num1, num2)
            elif choice == '*':
                res = multiply(num1, num2)
            elif choice == '/':
                res = divide(num1, num2)
            elif choice == '^':
                res = exponent(num1, num2)
            elif choice == '%':
                res = modulo(num1, num2)
                
            operation_str = f"{num1} {choice} {num2} = {res}"
            print(f"Result: {operation_str}")
            history.append(operation_str)
            
    except ValueError as e:
        if "could not convert string to float" in str(e):
            print("❌ Error: Please enter valid numeric values.")
        else:
            print(f"❌ {e}")
    except ZeroDivisionError as e:
        print(f"❌ {e}")
        
    print("-" * 40)
```
